```python
import urllib.request

# Read text from an online file
url = "https://raw.githubusercontent.com/hansotTLU/Digihum-Teh_2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/notebooks/assignment/text_mustang.txt"
text = urllib.request.urlopen(url).read().decode("utf-8")

# Print the first 187 characters
print("First 187 characters of the file:")
print(text[:187])

# Split into sentences
sentences = text.split("\n")
print("\nTotal number of sentences:", len(sentences))

# Split into words
words = text.split()
print("\nTotal number of words:", len(words))

# Print the first 12 words
print("\nFirst 12 words:")
print(words[:12])
# Prettier print:
print(*words[:12])

# Select only words longer than 10 letters
long_words = [w for w in words if len(w) > 10]

print("\nWords longer than 10 letters (first 10):")
print(long_words[:10])
# Prettier print:
print(*long_words[:10])

# Extract every 3rd word
every_third = words[::3]
print("\nEvery 3rd word (first 8):")
print(every_third[:8])
# Prettier print:
print(*every_third[:8])
```

```python
from wordcloud import WordCloud
import urllib.request

# Download text from URL
text_url = "https://raw.githubusercontent.com/hansotTLU/Digihum-Teh_2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/notebooks/assignment/text_mustang.txt"  # Moby Dick
text = urllib.request.urlopen(text_url).read().decode("utf-8")

# Choose some stopwords
stopwords = ["the", "and", "of", "to", "a", "in", "is", "that", "its", "for", "as", "up", "down", "from", "out"]

# Generate a word cloud
wc = WordCloud(
    width=1920,
    height=1080,
    background_color="darkred",
    stopwords=stopwords
).generate(text)

# Display the image
image = wc.to_image()
image.show()
```

## WordCloud description

The big words in the word cloud show that the text focuses mainly on the "Mustang", second most popular words are "performance", "American", "muscle", "iconic" and "Ford". The smaller words show that they weren't as popular in the text, for example: "high", "automotive", "turbo", "stand" and others.
