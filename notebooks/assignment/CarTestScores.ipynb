**What car was tested the most?**

Tesla Model 3 (nickname Nova) was tested the most on average, showing the highest test distance, while Honda Civic (nickname Ranger) and Toyota Corolla (Strider) were tested the least.

```python
# What car was tested the most? Bar chart
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("CarTestScores.csv")
effort=df.groupby("Nickname") ["Test_Distance_km"].mean().sort_values(ascending=False)

effort.plot(kind="bar", x="Nickname", y="Test_Distance_km", legend=False)
plt.ylabel(",Test_Distance_km")
plt.title("What car was tested the most?")
plt.show()
```

**What car performed the best?**

Tesla Model 3 (Nova) achieved the highest overall performance scores, clearly outperforming all other cars, while the Honda Civic (Ranger) had the lowest average performance.

```python
# What car performed the best? Bar chart

effort.plot(kind="bar", x="Nickname", y="Performance_Score", legend=False)
plt.ylabel("Performance_Score")
plt.title("What car performed the best?")
plt.show()
```

**What car performed the best?**

The pie chart shows that Nova (Tesla Model 3) dominates in performance contribution, indicating it consistently scores higher than others, whereas Ranger and Strider contribute much less to the total performance share. However, the results aren't too far apart.

```python
# What car performed the best? Pie chart

effort=df.groupby("Nickname") ["Performance_Score"].mean()

plt.figure(figsize=(8, 8))

effort.plot(kind="pie", x="Nickname", y="Performance_Score", legend=False, autopct='%1.1f%%')
plt.ylabel("Performance_Score")
plt.title("What car performed the best?")
plt.show()
```

**Does more testing = better score?**

The scatter plot suggests a weak positive correlation between testing distance and performance — cars tested over longer distances tend to have slightly higher performance scores, though it is not perfectly linear.

```python
# Does more testing = better score? Scatter

plt.scatter(df["Test_Distance_km"], df["Performance_Score"], s=df["Test_Distance_km"]*2, alpha=0.6)
plt.xlabel("Test_Distance_km")
plt.ylabel("Performance_Score")
plt.title("Does more testing = better score?")
plt.show()
```

**How consistent are test patterns and performance across cars?**

The boxplot shows that both Test_Distance_km and Performance_Score have moderate variation, but no extreme outliers — indicating relatively consistent testing efforts and performance results across all vehicles.

```python
# How consistent are test patterns and performance across cars? Boxplot

km_min=df["Test_Distance_km"].min()
km_q1=df["Test_Distance_km"].quantile(0.25)
km_q2=df["Test_Distance_km"].quantile(0.75)
km_max=df["Test_Distance_km"].max()
print("Minimum km tested:", km_min, "\nQ1:", km_q1)
print("Q2:", km_q2, "\nMaximum km tested:", km_max)

plt.Figure(figsize=(8, 6))
df[["Test_Distance_km", "Performance_Score"]].boxplot()
plt.ylabel("Value")
plt.title("How consistent are test patterns and performance across cars?")
plt.show()
```
