```python
# Who puts in the most effort? Bar chart
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("competencies.csv")
effort=df.groupby("Student") ["Hours_Studied"].mean().sort_values(ascending=False)

effort.plot(kind="bar", x="Student", y="Hours_Studied", legend=False)
plt.ylabel("Hours Studied")
plt.title("Who puts in the most effort?")
plt.show()
```

```python
# Who performs best? Bar chart
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("competencies.csv")
effort=df.groupby("Student") ["Grade"].mean().sort_values(ascending=False)

effort.plot(kind="bar", x="Student", y="Grade", legend=False)
plt.ylabel("Grade")
plt.title("Who performs best?")
plt.show()
```

```python
# Who performs best? Pie chart
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("competencies.csv")
effort=df.groupby("Student") ["Grade"].mean()

plt.figure(figsize=(8, 8))

effort.plot(kind="pie", x="Student", y="Grade", legend=False, autopct='%1.1f%%')
plt.ylabel("Grade")
plt.title("Who performs best?")
plt.show()
```

```python
# Does more studying = better grades?
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("competencies.csv")

plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*30, alpha=0.6)
plt.xlabel("Hours Studied")
plt.ylabel("Grade")
plt.title("Does more studying = better grades?")
plt.show()
```

```python
# How consistent are study patterns and performance? Boxplot
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("competencies.csv")

hours_min=df["Hours_Studied"].min()
hours_q1=df["Hours_Studied"].quantile(0.25)
hours_q2=df["Hours_Studied"].quantile(0.75)
hours_max=df["Hours_Studied"].max()
print("Minimum hours studied:", hours_min, "\nQ1:", hours_q1)
print("Q2:", hours_q2, "\nMaximum hours studied:", hours_max)

plt.Figure(figsize=(8, 6))
df[["Hours_Studied", "Grade"]].boxplot()
plt.ylabel("Value")
plt.title("How consistent are study patterns and performance?")
plt.show()
```
