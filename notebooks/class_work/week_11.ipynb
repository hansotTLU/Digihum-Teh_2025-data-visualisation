```python
sentence = "racecarlevelfield"
# [start:end]
# print first 3 letters
print(sentence[0:3])
print(sentence[:3])
# extract letters from 5-9
print(sentence[4:9])
# extract letters from 7th to end
print(sentence[7:])
# extract the last letter
print(sentence[-1])
# extract the last 4 characters
print(sentence[-4:])
# extract the first letter of the substring starting from 5
print(sentence[7:][0])
# extract the first 5 letters of the substring starting from 5
print(sentence[7:][:5])
# extract entire string/text
print(sentence[:])
# extract every 2nd character
print(sentence[::2])
```

```python
sentence = "the quick brown fox jumped over the lazy dog"
words = sentence.split(" ")
print(words)
# extract the first word
print(words[0])
# extract the first 4 words
print(words[:4])
print(*words[:4])
# extract the last 3 words
print(words[-3:])
print(*words[-3:])
# print number of words
print("number of words in the sentence: ", len(words))
```
