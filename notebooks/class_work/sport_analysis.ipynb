```python
import pandas as pd

df=pd.read_csv("sports.csv")
df.head() #.head() shows first rows, only df shows all
```

```python
df.describe() #generates some basic statistics
```

Box plot for each athlete to find the number of years of practise

```python
import matplotlib.pyplot as plt

df.plot(kind="bar", x="Name", y="Years of practice", legend=False) #generates a bar chart
plt.ylabel("Years")
plt.title("Years of practice per athlete")
plt.show()
```

```python
import matplotlib.pyplot as plt

df.plot(kind="bar", x="Name", y="Age", legend=False) #generates a bar chart
plt.ylabel("Age")
plt.title("Age of each athlete")
plt.show()
```

**Age vs Years of Practice**

- plt.scatter(x,y) - makes points for each pair of values
- "s" controls the size of each bubble
- s=df["Years of practice"]*30 - means if a student has more years of practice, their bubble is larger. *30 if a scaling factor (multiplier) to make the bubble big

```python
plt.scatter(df["Age"], df["Years of practice"], s=df["Years of practice"]*30, alpha=0.6)
plt.xlabel("Age")
plt.ylabel("Years")
plt.title("Age vs Years of Practice")
plt.show()
```

**Average age per sport**

```python
avg_age=df.groupby("Sport") ["Age"].mean()
avg_age.plot(kind="bar")
plt.ylabel("Age")
plt.title("Average age per sport")
plt.show()
```
